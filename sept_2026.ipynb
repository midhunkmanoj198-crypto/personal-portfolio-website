```python
#wap to read a number from the user and check it is prime or not

n=int(input("enter the number"))

def prime(n):
  c=0
  for i in range (1,n+1):
    if n%i==0:
      c=c+1

  if c==2:
    return "prime"
  else:
    return "not prime"
prime(n)
```

```python
#Recurssion : a function calling itself

#wap to find factorial of a number using recurssion

n=int(input("enter the no"))
def fac(n):
  if n==1:
    return 1
  else:
    return n*fac(n-1)

fac(n)
```

```python
#wap to find sum of first n natural no using recurssion

n=int(input())

def s(n):
  if n==1:
    return 1
  else:
    return n+s(n-1)
s(n)
```

```python
#Create a class of rectangles and create methods to display its area and perimeter

class Rectangle():
  def get_data(self):
    self.length=int(input("enter the length: "))
    self.length=int(input("enter the bredth: "))
  def display_data(self):
    print(self.length)
    print(self.bredth)

  def area(self):
    print(self.length*self.bredth)

  def perimeter(self):
    print(2*(self.length+self.bredth))

r1=Rectangle()
r1.get_data()
r1.display_data()
r1.area()
r1.perimeter()
```

```python
#Create a class of circle and print area and perimeter using methods

class Circle():
  def get_data(self):
    self.r=int(input("Enter the radius"))

def display_data(self):
  print(self.r)

def area(self):
  print(3.14*self.r*self.r)

def perimeter(self):
  print(2*3.14*self.r)

c1=Circle()

c1.get_data()
c1.display_data()
c1.area()
c1.perimeter()
```

```python
#student
 #name
 #rollno
  #get_data
  #display_data()

class Student():
  def __init__(self,n,r):
    self.name=n
    self.roll_no=r
  def display_data(self):
    print(self.name)
    print(self.roll_no)
#test
 #tmark
  #get_tmark()
  #display_tmark()
class Test(Student):
  def __init__(self,n,r,t):
    Student.__init__(self,n,r)
    self.tmark=t
  def display_tmark(self):
    print(self.tmark)

#sports
 #smark
 #display_smark()
class Sports():

  def __init__(self,s):
    self.smark=s
  def display_smark(self):
    print(self.smark)

#grade(test,sports)
 #tomark=tmark+smark
 #calgrade()
 #tomark>80  : Grade A
 #tomark>50 : Grade B
 #else : failed

class Grade(Test,Sports):

  def __init__(self,n,r,t,s):
    Sports.__init__(self,s)
    Test.__init__(self,n,r,t,)

  def calgrade(self):
    self.tomark=self.tmark+self.smark

    if self.tomark>80:
      print("Grade A")
    elif self.tomark>50:
      print("Grade B")
    else:
      print("failed")

g1=Grade("abu",42,40,20)
#g1.get_smark()
#g1.get_tmark()
g1.calgrade()
```

```python
#Method overriding

class Addition():
  def __init__(self,n):
    self.value=n
  def __add__(self,other):
    return self.value+other.value


a1=Addition(10)
a2=Addition(12)

print(a1+a2)

```
